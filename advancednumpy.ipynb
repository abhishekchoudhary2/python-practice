```python
import numpy as np
```

```python
a=np.random.randint(1,100,(10,10))
print(a)
print(a.shape)
print(a.size)
print(a.dtype)
print(a.itemsize)
print(a.ndim)
```

```python
a=np.arange(20)
r=a.reshape(4,5)
rr=a.reshape(10,2)
print(r)
print(rr)
```

```python
a=np.random.randn(5,5)
a.astype(float)
a[a<0.5]=0
print(a)
```

```python
a=np.random.randint(1,101,100)
print(a)
u=np.unique(a)
print(u)
```

```python
a=np.eye(7)
print(a)
a[np.diag_indices(7)]=np.random.randint(1,10,7)
print(a)
```

```python
dices=np.random.randint(1,7,10)
print("first array:",dices)
values,frequencies=np.unique(dices,return_counts=True)
print("frequencies:")
for value,frequency in zip(values,frequencies):
    print(value,":",frequency)
```

```python
a=np.random.randint(1,50,20)
print("array:",a)
parts=np.split(a,4)
print("subarray:")
for part in parts:
    print(part)
```

```python
a=np.random.randint(1,20,(6,6))
print("elements:",a)
n=a[a>10]
print("greater than:",n)
```

```python
a=np.array([10,20,30,40,50,60])
print("array:",a)
v=a.view()
print("slice:",a[0:4])
a[0]=-1
print("view:",a)
c=a.copy()
a[0]=100
print("copy:",a)
```

```python
a=np.random.randint(0,9,(3,3,3))
print(a.shape)
r=a.reshape(9,3)
print(r)
```

```python
a=np.arange(1,101).reshape(10,10)
border=np.concatenate([
a[0,:],
a[-1,:],
a[1:-1,0],
a[1:-1,-1]
])
print("array:",a)
print("border:",border)
```

```python
a=np.arange(1,101).reshape(10,10)
row=np.arange(10)
cols=np.arange(10)
diagonal=a[row,cols]
print(a)
print("diagonal:",diagonal)
```

```python
a=np.arange(64).reshape(4,4,4)
i,j,k=np.indices(a.shape)
result=a[(i+j+k)%2==0]
print(result)
```
