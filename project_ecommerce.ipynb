```python
# Q9. Display the top 10 orders based on Sales. Show only Order_ID, Customer_Name, Product, Quantity, and Sales.
df.sort_values(by="Sales",ascending = False)[["Order_ID","Customer_Name","Product","Quantity","Sales"]].head(10)
```

```python
# Q10. Find all orders where the discount is greater than 20%. Sort the result by Discount from highest to lowest.
df[df["Discount"]>0.20].sort_values(by = "Discount" ,ascending = False)
```

```python
# Q11. Find the 10 products with the highest average rating. Display the product name and average rating.
df.groupby("Product")["Rating"].mean().sort_values(ascending = False).head(10)
```

```python
# Q12. Calculate the total sales generated by each Category. Sort the result from highest to lowest sales.
df.groupby("Category")["Sales"].sum().sort_values(ascending = False)
```

```python
# Q13. Find the total quantity sold and total sales for each Product. Display the top 10 products based on total sales.
df.groupby("Product").agg({"Quantity":"sum","Sales":"sum"}).sort_values(by = "Sales",ascending = False).head(10)
```

```python
# Q14. Calculate the average Sales, average Rating, and average Delivery_Days for each Payment_Mode
df.groupby("Payment_Mode").agg({"Sales":"mean","Rating":"mean","Delivery_Days":"mean"})
```

```python
# Q15. Find the number of orders and total sales generated by each City. Which 5 cities have generated the highest sales?
df.groupby("City").agg({"Order_ID":"size","Sales":"sum"}).sort_values(by = "City",ascending=False)
```

```python
# Q16. Create a new column called Month from Order_Date. Find the total sales for each month.
df["Order_Date"]=pd.to_datetime(df["Order_Date"],errors = 'coerce')
df.groupby("Month")["Sales"].sum()
```

```python
df["Month"] = df["Order_Date"].dt.month_name()
```

```python
# Q17. Find the total sales for each Order_Status. Also calculate what percentage of total orders belong to each status.
status_summary = df.groupby("Order_Status").agg(
    total_sale = ("Sales","sum"),
    order_count = ("Sales","count"))
status_summary["Percentage"] = (status_summary["order_count"]/len(df))*100
status_summary
```

```python
# Q18. Compare the performance of different categories by calculating Total Sales, Average Sales, Total Quantity Sold, Average Rating, and Average Delivery Days. Which category appears to perform best overall?
category_performance = df.groupby("Category").agg(
    total_sales = ("Sales","sum"),
    average_sales =("Sales","mean"),
    sold_quantity = ("Sales","count"),
    average_rating = ("Rating","mean"),
    av_delivery_days = ("Delivery_Days","mean")
).reset_index()
category_performance
```
