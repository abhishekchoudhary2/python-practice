```python


data = {
    'Customer_Name': ['Rahul Sharma', 'Priya Singh', 'Amit Kumar', 'Neha Verma', 'Rohan Mehta', 'Anjali Gupta'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Sales': [25000, 800, 55000, 15000, 500, 40000],
    'Payment_Method': ['Credit Card', 'Cash', 'Debit Card', 'Credit Card', 'Debit Card', 'Cash'],
    'City': ['Jaipur', 'Delhi', 'Mumbai', 'Jaipur', 'Delhi', 'Mumbai']
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Sales_category"] = df["Sales"].apply( lambda x :"High" if x > 20000 else "low")
```

```python
df["Gender_code"] = df["Gender"].map({"Male":1,"Female":0})
```

```python
df
```

```python
df["Payment_Method"] = df["Payment_Method"].replace({"Credit Card" : "Card" , "Debit Card" :"Card"})
```

```python
df["sales_capped"] = df["Sales"].where(df["Sales"] <= 50000,50000)
```
