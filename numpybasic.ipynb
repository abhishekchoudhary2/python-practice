```python
a=np.arange(1,10).reshape(3,3)
print("array:",a)
rc=a[1,2]
print(rc)
```

```python
a=np.arange(1,10).reshape(3,3)
print("array:",a)
rc=a[1]
print("row:",rc)
```

```python
a=np.arange(1,10).reshape(3,3)
print("array:",a)
rc=a[:,2]
print("col:",rc)
```

```python
a=np.arange(1,10).reshape(3,3)
print("array:",a)
submatrix=a[0:2,0:2]
print("first two row and two column:",submatrix)
```
