```python
import pandas as pd
```

```python
import numpy as np
```

```python
sales_jan = pd.DataFrame({
    'Order_ID': [101, 102, 103],
    'Customer': ['Amit', 'Priya', 'Rahul'],
    'Sales': [5000, 7000, 6000]
})
sales_jan
```

```python
sales_feb = pd.DataFrame({
    'Order_ID': [104, 105, 106],
    'Customer': ['Neha', 'Vikas', 'Sneha'],
    'Sales': [8000, 4000, 9000]
})

sales_feb

```

```python
sales_mar = pd.DataFrame({
    'Order_ID': [107, 108, 109],
    'Customer': ['Karan', 'Riya', 'Mohit'],
    'Sales': [5500, 6500, 7500],
    'City': ['Delhi', 'Mumbai', 'Jaipur']
})

sales_mar
```

```python
df_sales = pd.concat([sales_jan,sales_feb],axis = 0)
print(df_sales)
```

```python
df_sales = pd.concat([sales_jan,sales_feb],ignore_index = True)
print(df_sales)
```

```python
df_sales = pd.concat([sales_jan,sales_feb,sales_mar],axis = 0)
print(df_sales)
```

```python
df_sales = pd.concat([sales_jan,sales_feb],axis = 1)
print(df_sales)
```

```python
pd.concat([sales_jan[["Customer"]], sales_feb[["Customer"]]],axis = 1)
```

```python
pd.concat([sales_jan[["Customer"]], sales_feb[["Customer"]]],axis = 1,keys = ["jan","feb"])
```

```python
df_sales = pd.concat([sales_jan,sales_feb,sales_mar])["Sales"].sum()
print(df_sales)
```
