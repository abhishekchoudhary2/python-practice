```python
import pandas as pd 
```

```python
import numpy as np
```

```python
data = {
    'Order_ID': [201, 202, 203, 204, 205, 206, 207, 208, 209, 210,
                 211, 212, 213, 214, 215],

    'Region': ['North', 'South', 'East', 'West', 'North',
               'South', 'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South', 'East'],

    'Product': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile',
                'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop',
                'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet'],

    'Salesperson': ['Raj', 'Priya', 'Aman', 'Neha', 'Raj',
                    'Priya', 'Aman', 'Neha', 'Raj', 'Priya',
                    'Aman', 'Neha', 'Raj', 'Priya', 'Aman'],

    'Sales': [55000, 30000, 22000, 48000, 35000,
              25000, 60000, 28000, 24000, 52000,
              33000, 27000, 65000, 31000, 21000],

    'Quantity': [2, 3, 2, 1, 4,
                 3, 2, 2, 3, 1,
                 5, 2, 2, 4, 3]
}

df = pd.DataFrame(data)

df
```

```python
pd.pivot_table(
    df,
    values = "Sales",
    index = "Region",
    columns = "Product",
    aggfunc = "sum"
)
```

```python
pd.pivot_table(
    df,
    values = "Sales",
    index = "Region",
    columns= "Product",
    aggfunc = "mean"
)
```

```python
pd.pivot_table(
    df,
    index = "Salesperson",
    columns = "Product",
    aggfunc = "sum"
)
```

```python
pd.pivot_table(
    df,
    values = "Sales",
    index = "Salesperson",
    columns = "Region",
    aggfunc = "sum"
)
```

```python
pd.pivot_table(
    df,
    values = "Sales",
    columns = "Salesperson",
    aggfunc = "mean"
)
```

```python
pd.pivot_table(
    df,
    values = "Quantity",
    index = "Salesperson",
    aggfunc = "sum"
)
```

```python
pd.pivot_table(
    df,
    values = "Sales",
    index = "Salesperson",
    aggfunc =["sum","mean"]
)
```

```python
pd.pivot_table(
    df,
    values = "Sales",
    index = ["Salesperson","Region"],
    columns = "Product",
    aggfunc = "sum"
)
```

```python
pd.pivot_table(
    df,
    values ="Sales",
    index = "Region",
    aggfunc = "sum"
).sort_values(by = "Sales",ascending = False)
```

```python
pd.pivot_table(
    df,
    values = "Quantity",
    index = "Product",
    aggfunc = "sum"
).sort_values(by="Quantity",ascending= False)
```
