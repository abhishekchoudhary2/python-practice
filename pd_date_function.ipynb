```python
df["DOB"] =pd.to_datetime (df ["DOB"] )
```

```python
df["DOB_Year"] = df ["DOB"] .dt.year
print(df[["Student_Name", "DOB_Year"]])
```

```python
df["Month_Number"] = df["DOB"].dt.month
df["Month_day"] = df["DOB"].dt.day
print(df[["Student_Name","Month_Number" , "Month_day"]] )
```

```python
df["day_name"] = df["DOB"].dt.day_name()
df["month_name"] = df["DOB"].dt.month_name()
print(df [["Student_Name" , "day_name" , "month_name" ]] )
```
